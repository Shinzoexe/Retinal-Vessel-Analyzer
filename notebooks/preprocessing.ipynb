```python
import cv2
import numpy as np

def preprocess_fundus(img):

    # Resize
    img = cv2.resize(img, (512, 512))

    # Green channel extraction
    green = img[:, :, 1]

    # CLAHE enhancement
    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8,8)
    )

    enhanced = clahe.apply(green)

    # Normalization
    enhanced = cv2.normalize(
        enhanced,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    return enhanced
```

```python
from tqdm import tqdm
import os

for file in tqdm(os.listdir(DRIVE_IMG)):

    img_path = os.path.join(DRIVE_IMG, file)

    img = cv2.imread(img_path)

    processed = preprocess_fundus(img)

    save_path = os.path.join(
        PROCESSED_DIR,
        "DRIVE/images",
        file.replace(".tif",".png")
    )

    cv2.imwrite(save_path, processed)

print("DRIVE Images Processed")
```

```python
for file in tqdm(os.listdir(DRIVE_MASK)):

    mask_path = os.path.join(DRIVE_MASK, file)

    mask = cv2.imread(mask_path, 0)

    mask = cv2.resize(mask, (512,512))

    save_path = os.path.join(
        PROCESSED_DIR,
        "DRIVE/masks",
        file.replace(".gif",".png")
    )

    cv2.imwrite(save_path, mask)

print("DRIVE Masks Processed")
```

```python
import os
import shutil
import random

random.seed(42)

FINAL_DIR = "/content/drive/MyDrive/dataset_final"
PROCESSED_DIR = "/content/drive/MyDrive/processed"

all_data = []

# DRIVE
for img in os.listdir(f"{PROCESSED_DIR}/DRIVE/images"):
    all_data.append(("DRIVE", img))

# HRF
for img in os.listdir(f"{PROCESSED_DIR}/HRF/images"):
    all_data.append(("HRF", img))

random.shuffle(all_data)

total = len(all_data)

train_size = int(0.70 * total)
val_size = int(0.15 * total)

train_data = all_data[:train_size]
val_data = all_data[train_size:train_size+val_size]
test_data = all_data[train_size+val_size:]

print("Train:", len(train_data))
print("Val  :", len(val_data))
print("Test :", len(test_data))
```
